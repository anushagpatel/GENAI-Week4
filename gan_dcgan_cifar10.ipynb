```


# ============================================================================
# MERGED FROM ORIGINAL NOTEBOOK CELLS — RUN THIS SINGLE CELL IN GOOGLE COLAB
# ============================================================================

import os, time, random
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, Model

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print('TensorFlow:', tf.__version__)
print('GPU(s):', tf.config.list_physical_devices('GPU'))

CLASS_NAMES = ['airplane','automobile','bird','cat','deer','dog','frog','horse','ship','truck']
CLASS_ID = 1                 # one class only: automobile
LATENT_DIM = 100
BATCH_SIZE = 128
EPOCHS = 100                 # 100 is a reasonable start; increase for better images
LEARNING_RATE = 2e-4
BETA_1 = 0.5
OUTPUT_DIR = 'dcgan_cifar10_outputs'
os.makedirs(OUTPUT_DIR, exist_ok=True)
print('Training class:', CLASS_NAMES[CLASS_ID])

# Load CIFAR-10 and keep only the selected class so the GAN learns one image category.
(x_train, y_train), (_, _) = tf.keras.datasets.cifar10.load_data()
y_train = y_train.squeeze()
x_train = x_train[y_train == CLASS_ID].astype('float32')
# tanh generator output requires real images scaled to [-1, 1]
x_train = (x_train - 127.5) / 127.5

print('Images in selected class:', x_train.shape)

fig, axes = plt.subplots(2, 5, figsize=(10, 4))
for ax, img in zip(axes.flat, x_train[:10]):
    ax.imshow((img + 1) / 2)
    ax.axis('off')
fig.suptitle(f'Real CIFAR-10 samples: {CLASS_NAMES[CLASS_ID]}')
plt.tight_layout()
plt.show()

# Build an efficient shuffled/batched input pipeline; prefetch overlaps data loading with training.
dataset = (tf.data.Dataset.from_tensor_slices(x_train)
           .shuffle(len(x_train), seed=SEED, reshuffle_each_iteration=True)
           .batch(BATCH_SIZE, drop_remainder=True)
           .prefetch(tf.data.AUTOTUNE))

# Builds the Generator: converts a random latent/noise vector into a 32x32 RGB image.
# Dense first creates a small 4x4 feature map; Conv2DTranspose layers progressively upsample it.
def build_generator(latent_dim=LATENT_DIM):
    z = layers.Input(shape=(latent_dim,), name='noise')
    x = layers.Dense(4 * 4 * 256, use_bias=False)(z)
    x = layers.BatchNormalization()(x)
    x = layers.LeakyReLU(negative_slope=0.2)(x)
    x = layers.Reshape((4, 4, 256))(x)

    x = layers.Conv2DTranspose(128, 4, strides=2, padding='same', use_bias=False)(x)  # 8x8
    x = layers.BatchNormalization()(x)
    x = layers.LeakyReLU(negative_slope=0.2)(x)

    x = layers.Conv2DTranspose(64, 4, strides=2, padding='same', use_bias=False)(x)   # 16x16
    x = layers.BatchNormalization()(x)
    x = layers.LeakyReLU(negative_slope=0.2)(x)

    out = layers.Conv2DTranspose(3, 4, strides=2, padding='same',
                                 activation='tanh', name='generated_image')(x)       # 32x32x3
    return Model(z, out, name='generator')

generator = build_generator()
generator.summary()

# Builds the Discriminator: takes a 32x32 RGB image and predicts whether it is real or generated.
# Strided Conv2D layers progressively downsample the image while learning visual features.
def build_discriminator():
    img = layers.Input(shape=(32, 32, 3), name='image')
    x = layers.Conv2D(64, 4, strides=2, padding='same')(img)       # 16x16
    x = layers.LeakyReLU(negative_slope=0.2)(x)
    x = layers.Dropout(0.3)(x)

    x = layers.Conv2D(128, 4, strides=2, padding='same', use_bias=False)(x)  # 8x8
    x = layers.BatchNormalization()(x)
    x = layers.LeakyReLU(negative_slope=0.2)(x)
    x = layers.Dropout(0.3)(x)

    x = layers.Conv2D(256, 4, strides=2, padding='same', use_bias=False)(x)  # 4x4
    x = layers.BatchNormalization()(x)
    x = layers.LeakyReLU(negative_slope=0.2)(x)
    x = layers.Dropout(0.3)(x)

    x = layers.Flatten()(x)
    out = layers.Dense(1)(x)  # logits; no sigmoid needed with from_logits=True
    return Model(img, out, name='discriminator')

discriminator = build_discriminator()
discriminator.summary()

# Binary cross-entropy is the adversarial objective used by both networks.
# Adam with beta_1=0.5 is a common DCGAN optimizer configuration.
bce = tf.keras.losses.BinaryCrossentropy(from_logits=True)
g_optimizer = tf.keras.optimizers.Adam(LEARNING_RATE, beta_1=BETA_1)
d_optimizer = tf.keras.optimizers.Adam(LEARNING_RATE, beta_1=BETA_1)

# Computes discriminator loss. It rewards predictions close to 1 for real images
# and close to 0 for images produced by the generator.
def discriminator_loss(real_logits, fake_logits):
    real_loss = bce(tf.ones_like(real_logits), real_logits)
    fake_loss = bce(tf.zeros_like(fake_logits), fake_logits)
    return real_loss + fake_loss

# Computes generator loss. The generator succeeds when the discriminator
# classifies generated images as real (target label = 1).
def generator_loss(fake_logits):
    return bce(tf.ones_like(fake_logits), fake_logits)

# Performs one GAN training step on a batch of real images.
# 1) Sample random noise, 2) generate fake images, 3) score real/fake images,
# 4) calculate both losses, 5) backpropagate, and 6) update both networks.
# @tf.function compiles this function into a TensorFlow graph for faster GPU execution.
@tf.function
def train_step(real_images):
    batch_n = tf.shape(real_images)[0]
    noise = tf.random.normal([batch_n, LATENT_DIM])
    with tf.GradientTape() as g_tape, tf.GradientTape() as d_tape:
        fake_images = generator(noise, training=True)
        real_logits = discriminator(real_images, training=True)
        fake_logits = discriminator(fake_images, training=True)
        g_loss = generator_loss(fake_logits)
        d_loss = discriminator_loss(real_logits, fake_logits)
    g_grads = g_tape.gradient(g_loss, generator.trainable_variables)
    d_grads = d_tape.gradient(d_loss, discriminator.trainable_variables)
    g_optimizer.apply_gradients(zip(g_grads, generator.trainable_variables))
    d_optimizer.apply_gradients(zip(d_grads, discriminator.trainable_variables))
    return g_loss, d_loss

fixed_noise = tf.random.normal([25, LATENT_DIM], seed=SEED)

# Generates images from supplied noise vectors and displays them as a grid.
# Generator output is converted from [-1, 1] back to [0, 1] for visualization.
# The optional save_path also saves the grid for reports/slides.
def show_generated(model, noise, title='', save_path=None, rows=5, cols=5):
    imgs = model(noise, training=False).numpy()
    imgs = np.clip((imgs + 1.0) / 2.0, 0, 1)
    fig, axes = plt.subplots(rows, cols, figsize=(cols * 1.7, rows * 1.7))
    axes = np.array(axes).reshape(-1)
    for ax, img in zip(axes, imgs):
        ax.imshow(img)
        ax.axis('off')
    for ax in axes[len(imgs):]: ax.axis('off')
    fig.suptitle(title)
    plt.tight_layout()
    if save_path: plt.savefig(save_path, dpi=150, bbox_inches='tight')
    plt.show()

# Before training: useful baseline for slides
show_generated(generator, fixed_noise, 'Before training')

g_history, d_history, epoch_seconds = [], [], []

# Main training loop: train on every batch, record losses/timing, and periodically
# generate the same fixed-noise samples so training progress can be compared visually.
for epoch in range(1, EPOCHS + 1):
    start = time.perf_counter()
    g_metric = tf.keras.metrics.Mean()
    d_metric = tf.keras.metrics.Mean()
    for real_batch in dataset:
        g_loss, d_loss = train_step(real_batch)
        g_metric.update_state(g_loss)
        d_metric.update_state(d_loss)

    g_history.append(float(g_metric.result()))
    d_history.append(float(d_metric.result()))
    epoch_seconds.append(time.perf_counter() - start)

    if epoch == 1 or epoch % 10 == 0:
        print(f'Epoch {epoch:03d}/{EPOCHS} | G={g_history[-1]:.4f} | D={d_history[-1]:.4f} | {epoch_seconds[-1]:.2f}s')
    if epoch % 25 == 0 or epoch == EPOCHS:
        show_generated(generator, fixed_noise,
                       f'{CLASS_NAMES[CLASS_ID]} — epoch {epoch}',
                       f'{OUTPUT_DIR}/epoch_{epoch:03d}.png')

generator.save(f'{OUTPUT_DIR}/generator.keras')
discriminator.save(f'{OUTPUT_DIR}/discriminator.keras')

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(g_history, label='Generator loss')
ax.plot(d_history, label='Discriminator loss')
ax.set_xlabel('Epoch')
ax.set_ylabel('Loss')
ax.set_title('DCGAN training losses')
ax.legend()
plt.show()

print(f'Mean epoch time: {np.mean(epoch_seconds):.3f} s')
print(f'Total training time: {np.sum(epoch_seconds)/60:.2f} min')
print(f'Generator parameters: {generator.count_params():,}')
print(f'Discriminator parameters: {discriminator.count_params():,}')

# Creates intermediate latent vectors between two noise vectors z_a and z_b.
# This demonstrates that smoothly changing the noise can produce smooth changes
# in generated images, revealing structure learned in the latent space.
def interpolate_noise(z_a, z_b, steps=10):
    alphas = np.linspace(0.0, 1.0, steps, dtype=np.float32)
    return tf.stack([(1-a)*z_a + a*z_b for a in alphas])

z_a = tf.random.normal([LATENT_DIM], seed=101)
z_b = tf.random.normal([LATENT_DIM], seed=202)
z_interp = interpolate_noise(z_a, z_b, steps=10)
show_generated(generator, z_interp,
               'Controlled noise: interpolation zA → zB',
               f'{OUTPUT_DIR}/noise_interpolation.png', rows=1, cols=10)

# Keeps every noise coordinate fixed except one selected dimension.
# Sweeping that coordinate from negative to positive values helps show whether
# a particular latent dimension controls a visible feature of the generated image.
def latent_dimension_traversal(base_z, dim=0, values=None):
    if values is None:
        values = np.linspace(-3, 3, 9, dtype=np.float32)
    zs = np.repeat(np.asarray(base_z, dtype=np.float32)[None, :], len(values), axis=0)
    zs[:, dim] = values
    return tf.convert_to_tensor(zs), values

base_z = np.random.default_rng(SEED).normal(size=LATENT_DIM).astype(np.float32)
z_sweep, sweep_values = latent_dimension_traversal(base_z, dim=0)
show_generated(generator, z_sweep,
               'One-coordinate traversal: z[0] = -3 … +3',
               f'{OUTPUT_DIR}/noise_dimension_0.png', rows=1, cols=len(sweep_values))
print('Values used:', sweep_values)

# Noise-control experiment 3: keep the latent direction fixed and only change
# its magnitude. This shows how distance from the latent-space origin affects output.
direction = tf.random.normal([LATENT_DIM], seed=303)
scales = np.linspace(0.0, 2.0, 9, dtype=np.float32)
z_scaled = tf.stack([float(s) * direction for s in scales])
show_generated(generator, z_scaled,
               'Same latent direction, different noise magnitude',
               f'{OUTPUT_DIR}/noise_magnitude.png', rows=1, cols=len(scales))
print('Scales used:', scales)

# Save key performance statistics to CSV so this implementation can later be
# compared fairly with the Claude-generated implementation.
import csv
summary_path = f'{OUTPUT_DIR}/chatgpt_dcgan_summary.csv'
with open(summary_path, 'w', newline='') as f:
    w = csv.writer(f)
    w.writerow(['implementation','class','epochs','batch_size','latent_dim','generator_params','discriminator_params','mean_epoch_seconds','final_g_loss','final_d_loss'])
    w.writerow(['ChatGPT_DCGAN', CLASS_NAMES[CLASS_ID], EPOCHS, BATCH_SIZE, LATENT_DIM,
                generator.count_params(), discriminator.count_params(),
                np.mean(epoch_seconds), g_history[-1], d_history[-1]])
print('Saved:', summary_path)
```
